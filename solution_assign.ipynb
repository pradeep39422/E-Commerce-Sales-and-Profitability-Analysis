```python
import pandas as pd
import numpy as np
```

```python
orders_df = pd.read_excel('List_of_Orders.xlsx')
details_df = pd.read_excel('Order_Details.xlsx')
target_df = pd.read_excel('Sales_target.xlsx')
```

```python
orders_df.head()
```

```python
merged_df = pd.merge(orders_df, details_df, on='Order ID')
```

```python
category_metrics = merged_df.groupby('Category').agg(
    Total_Sales=('Amount', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Unique_Order_Count=('Order ID', 'nunique')
).reset_index()
```

```python
category_metrics['Avg_Profit_Per_Order'] = category_metrics['Total_Profit'] / category_metrics['Unique_Order_Count']
category_metrics['Profit_Margin_%'] = (category_metrics['Total_Profit'] / category_metrics['Total_Sales']) * 100

print("--- Part 1: Category Metrics ---")
print(category_metrics)
```

***Part 1: Sales and Profitability Analysis***

---



Top-performing Category: Clothing performed the best overall, with the highest profit margin (8.03%) and the most orders (393). Although Electronics generated the highest sales (₹165,267) and the highest profit per order (₹51.44), Clothing delivered more consistent and profitable results.

Underperforming Category: Furniture was the weakest performer. Despite sales of ₹127,181, it earned only ₹2,298 in profit, giving it a low profit margin of 1.81% and an average profit of just ₹12.35 per order.

Reason: Furniture likely has higher costs for shipping, storage, and discounts, which reduce profits. Clothing, on the other hand, usually has lower costs and better margins, making it more profitable overall.

```python
state_orders = orders_df.groupby('State')['Order ID'].nunique().reset_index()
top_5_states = state_orders.sort_values('Order ID', ascending=False).head(5)['State'].tolist()

top_states_df = merged_df[merged_df['State'].isin(top_5_states)]

state_metrics = top_states_df.groupby('State').agg(
    Total_Sales=('Amount', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Unique_Order_Count=('Order ID', 'nunique')
).reset_index()

state_metrics['Avg_Profit_Per_Order'] = state_metrics['Total_Profit'] / state_metrics['Unique_Order_Count']

state_metrics = state_metrics.set_index('State').loc[top_5_states].reset_index()

print("--- Part 3: Regional Performance (Top 5 States) ---")
print(state_metrics[['State', 'Total_Sales', 'Avg_Profit_Per_Order']])
```

Priority Areas for Improvement:
